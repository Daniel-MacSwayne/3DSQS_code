```python
# ── Make fly-through video from interp/render frames ──────────────────────────
# Standalone — no dependency on Train.ipynb variables.
# Uses imageio + pyav (libx264) for proper H.264 MP4.
# Install once if needed: pip install imageio[pyav]

import os, imageio, cv2
import numpy as np

Vid_Dataset = "Deep_Blending"
Vid_Scene   = "Playroom"
Vid_FPS     = 30

# ── Paths ───────────────────────────────────────────────────────────────────
Project_Path = '/home/daniel/Documents/Projects/Sync/3D/'
frames_dir   = f'{Project_Path}Results/{Vid_Dataset}/{Vid_Scene}/SQE/interp/render/'
out_path     = f'{Project_Path}Results/{Vid_Dataset}/{Vid_Scene}/SQE/{Vid_Scene}_flythrough.mp4'

# ── Find frames ──────────────────────────────────────────────────────────────
if not os.path.exists(frames_dir):
    print(f"[ERROR] frames folder not found:\n  {frames_dir}")
else:
    files = sorted([
        f for f in os.listdir(frames_dir)
        if f.lower().endswith(('.png', '.jpg')) and os.path.isfile(os.path.join(frames_dir, f))
    ])
    print(f"Found {len(files)} frames  ({frames_dir})")

    if not files:
        print("[ERROR] No image files found.")
    else:
        # Load frames as RGB numpy arrays (imageio expects RGB, cv2 reads BGR)
        frames = []
        for f in files:
            img = cv2.imread(os.path.join(frames_dir, f))
            if img is not None:
                frames.append(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

        # Write H.264 MP4 via imageio + pyav (libx264)
        imageio.mimsave(out_path, frames, fps=Vid_FPS, codec='libx264')

        size_mb = os.path.getsize(out_path) / 1e6
        h, w    = frames[0].shape[:2]
        print(f"Saved: {out_path}")
        print(f"  {len(frames)} frames  @  {Vid_FPS} fps  |  {w}×{h}  |  {size_mb:.1f} MB")
```
